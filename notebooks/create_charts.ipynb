```python
import pandas as pd
import plotly.express as px


df = pd.read_csv("msft_analysis.csv")
```

```python
fig = px.line(
    df,
    x="Fiscal Year",
    y="Revenue",
    title="Microsoft Revenue (2021–2025)",
    markers=True,
)

fig.update_layout(
    template="plotly_white",
    yaxis_title="Revenue ($)",
)

fig.write_html("revenue.html")
fig.show()
```

```python
fig = px.bar(
    df,
    x="Fiscal Year",
    y="Net Income",
    title="Net Income",
    text_auto=".2s",
)

fig.update_layout(template="plotly_white")

fig.write_html("net_income.html")
fig.show()
```

```python
fig = px.bar(
    df,
    x="Fiscal Year",
    y="Free Cash Flow",
    title="Free Cash Flow",
    text_auto=".2s",
)

fig.update_layout(template="plotly_white")

fig.write_html("fcf.html")
fig.show()
```

```python
fig = px.line(
    df,
    x="Fiscal Year",
    y="Revenue Growth %",
    title="Revenue Growth (%)",
    markers=True,
)

fig.update_layout(template="plotly_white")

fig.write_html( "revenue_growth.html")
fig.show()
```

```python
margin_df = df[
    [
        "Fiscal Year",
        "Gross Margin %",
        "Operating Margin %",
        "Net Margin %",
    ]
].melt(
    id_vars="Fiscal Year",
    var_name="Metric",
    value_name="Percent",
)

fig = px.line(
    margin_df,
    x="Fiscal Year",
    y="Percent",
    color="Metric",
    markers=True,
    title="Profitability Margins",
)

fig.update_layout(template="plotly_white")

fig.write_html( "margins.html")
fig.show()
```

```python
# ROE (Return on Equity) and ROA (Return on Assets)
returns = df[
    [
        "Fiscal Year",
        "ROE %",
        "ROA %",
    ]
].melt(
    id_vars="Fiscal Year",
    var_name="Metric",
    value_name="Percent",
)

fig = px.bar(
    returns,
    x="Fiscal Year",
    y="Percent",
    color="Metric",
    barmode="group",
    title="ROE vs ROA",
)

fig.update_layout(template="plotly_white")

fig.write_html( "returns.html")
fig.show()
```
